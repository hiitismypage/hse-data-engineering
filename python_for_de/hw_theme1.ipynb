Задание 1. 
Переведите содержимое файла purchase_log.txt в словарь purchases вида:{‘1840e0b9d4’: ‘Продукты’, …}

```python
import json

purchases = {}

with open('data/purchase_log.txt', 'r', encoding='utf-8') as f:
    for line in f:
        row = json.loads(line)
        if row['user_id'] == 'user_id':
            continue
        purchases[row['user_id']] = row['category']

for key, value in list(purchases.items())[:2]:
    print(key, repr(value))
```

Задание 2 (необязательное).
Для каждого user_id в файле visit_log.csv определите третий столбец с категорией покупки. Если покупка была, сам файл visit_log.csv изменять не надо.
Запишите в файл funnel.csv визиты из файла visit_log.csv, в которых были покупки с указанием категории.


```python
import csv

with open('data/visit_log.csv', 'r', encoding='utf-8') as fin, \
     open('data/funnel.csv', 'w', encoding='utf-8', newline='') as fout:

    reader = csv.reader(fin)
    writer = csv.writer(fout)

    header = next(reader)
    writer.writerow(header + ['category'])

    for user_id, source in reader:
        category = purchases.get(user_id)
        if category is not None:
            writer.writerow([user_id, source, category])


with open('data/funnel.csv', 'r', encoding='utf-8') as f:
    for i in range(3):
        print(f.readline().strip())
```
